```python
def hosg(isim,dil="Türkçe",buyuk_harf=False):
  mesajlar={
      "Türkçe":f"Merhaba {isim}",
      "İngilizce":f"Hello {isim}",
      "Almanca":f"Hallo {isim}",
  }
  sonuc=mesajlar.get(dil,f"Hi {isim}")
  return sonuc.upper() if buyuk_harf else sonuc
print(hosg("Ahmet","İngilizce"))
print(hosg("kgfch","Almanca",True))
print(hosg("sdkjnavkjdnsa","",True))
```

```python
def alist():
  liste=[]
  print("Haftalık alışveriş listesine ürün ekleyin.")
  print("Çıkmak için 'q' yazın.")
  while True:
    urun=input("ürün ekleyin: ")
    if urun.lower()=="q":break
    liste.append(urun.strip())
    print(f"{urun} listeye eklendi\n")
  return liste
```
